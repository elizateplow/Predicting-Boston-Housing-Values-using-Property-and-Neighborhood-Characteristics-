### Part 2 - State Source and Load Data

**Data Source:** The data for this project comes from the City of Boston's 'Analyze Boston' data hub and contains property assessment information for properties in Boston. I accessed the data using the CKAN Data API provided on the website (https://data.boston.gov/dataset/property-assessment/resource/ee73430d-96c0-423e-ad21-c4cfb54c8961). 

```python
import urllib.request
import json
import pandas as pd

all_records = []
resource_id = "ee73430d-96c0-423e-ad21-c4cfb54c8961"
limit = 1000
offset = 0

while True:  
    url = (
    "https://data.boston.gov/api/3/action/datastore_search"
    f"?resource_id={resource_id}&limit={limit}&offset={offset}")

    with urllib.request.urlopen(url) as response:   
        data = json.load(response)

    records = data["result"]["records"]

    if len(records) == 0:
        break

    all_records.extend(records)

    offset += limit 

df = pd.DataFrame(all_records)
```

```python
df.head()
```

```python
print(df.shape)
print(df.columns)
```

```python
housing_df = df[["TOTAL_VALUE", "LIVING_AREA", "LAND_VALUE", "LAND_SF", "BED_RMS",
                 "FULL_BTH", "TT_RMS", "YR_BUILT", "YR_REMODEL", "NUM_PARKING",
                 "FIREPLACES", "CITY", "ZIP_CODE", "LU_DESC", "BLDG_TYPE",
                 "OVERALL_COND", "AC_TYPE", "HEAT_TYPE", "PROP_VIEW"]].copy()

housing_df.head()
```

```python
numeric = ["TOTAL_VALUE", "LIVING_AREA", "LAND_VALUE", "LAND_SF", "BED_RMS", 
           "FULL_BTH", "TT_RMS", "YR_BUILT", "YR_REMODEL", "NUM_PARKING", 
           "FIREPLACES"]

for x in numeric:
    housing_df[x] = housing_df[x].str.replace(",", "")
    housing_df[x] = pd.to_numeric(housing_df[x])
```

```python
categorical = ["CITY", "ZIP_CODE", "LU_DESC", "OVERALL_COND", "AC_TYPE", 
               "HEAT_TYPE", "BLDG_TYPE", "PROP_VIEW"]

for x in categorical:
    housing_df[x] = housing_df[x].astype("category")
```
